# Bước 2: Statistics – hiểu dữ liệu qua thống kê

Mục tiêu: trả lời các câu hỏi cơ bản về dữ liệu **trước khi** xử lý hay huấn luyện gì.
Mỗi ô bên dưới là một câu hỏi. Bấm **Run** từng ô theo thứ tự, đọc kết quả, rồi ghi nhận xét của bạn vào ô cuối.

Các hàm thống kê nằm ở `src/netflix/pipelines/statistics/nodes.py`; notebook này chỉ gọi lại chúng.

```python
import importlib.util
import sys
from pathlib import Path

# Kiểm tra kernel đã có đủ thư viện chưa, để báo lỗi dễ hiểu nếu thiếu
required = ["kedro", "kedro_datasets", "pandas", "pyarrow", "scipy", "sklearn"]
missing = [m for m in required if importlib.util.find_spec(m) is None]
if missing:
    raise ModuleNotFoundError(
        f"Kernel này ({sys.executable}) thiếu: {missing}.\n"
        "Chạy trong terminal rồi khởi động lại kernel:\n"
        f'  "{sys.executable}" -m pip install -r requirements.txt'
    )

import pandas as pd
from kedro.framework.session import KedroSession
from kedro.framework.startup import bootstrap_project

print("Python đang dùng:", sys.executable)  # phải nằm trong thư mục .venv


def find_project_root() -> Path:
    # Tìm thư mục gốc của dự án Kedro, dù notebook được mở từ đâu
    here = Path(globals().get("__vsc_ipynb_file__", Path.cwd() / "x")).resolve().parent
    for folder in (here, *here.parents, Path.cwd(), *Path.cwd().parents):
        if (folder / "conf").is_dir() and (folder / "pyproject.toml").exists():
            return folder
    for pyproject in Path.cwd().glob("*/pyproject.toml"):  # mở từ thư mục cha
        if "[tool.kedro]" in pyproject.read_text(encoding="utf-8"):
            return pyproject.parent
    raise FileNotFoundError("Không tìm thấy dự án Kedro (thư mục có conf/ và pyproject.toml)")


project_path = find_project_root()
print("Thư mục dự án:", project_path)

bootstrap_project(project_path)  # thêm thư mục src vào đường dẫn import

# Import sau bootstrap_project để chạy được dù gói netflix chưa được pip install
from netflix.pipelines.statistics.nodes import (  # noqa: E402
    churn_rate_by_feature,
    data_quality,
    describe_categorical,
    describe_numeric,
    target_association,
    target_distribution,
)

with KedroSession.create(project_path=project_path) as session:
    context = session.load_context()
    df = context.catalog.load("netflix_raw")  # dữ liệu gốc, khai báo trong catalog.yml
    params = context.params["statistics"]     # tham số trong parameters_statistics.yml

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 30)
print("Kích thước:", df.shape)
df.head()
```

## Câu hỏi 1: Dữ liệu có đầy đủ và sạch không?
Kiểm tra giá trị thiếu và dòng trùng. Nếu có thì ở bước tiền xử lý phải xử lý.

```python
data_quality(df, params)
```

## Câu hỏi 2: Hai lớp "rời" và "ở lại" có cân bằng không?
Nếu một lớp chiếm áp đảo (ví dụ 95%) thì chỉ số accuracy sẽ đánh lừa, và cần cách xử lý riêng.

```python
target_distribution(df, params)
```

## Câu hỏi 3: Các cột số trông như thế nào?
- `mean` và `median` lệch nhau nhiều hoặc `skew` lớn: phân phối bị lệch.
- `outliers_iqr`: số giá trị nằm quá xa phần lớn dữ liệu.

```python
describe_numeric(df, params)
```

## Câu hỏi 4: Các cột chữ có bao nhiêu mức, mức nào phổ biến?
Mức hiếm (`smallest_share` rất nhỏ) có thể gây khó cho mô hình. Ở đây các mức khá đều nhau.

```python
describe_categorical(df, params)
```

## Câu hỏi 5: Cột nào liên quan đến việc rời bỏ?
- `strength`: độ mạnh của mối liên hệ, từ 0 đến 1.
- `p_value` nhỏ hơn 0,05 (`significant = True`): mối liên hệ khó là do ngẫu nhiên.

Cột có `significant = False` gần như không giúp dự đoán churn.

```python
target_association(df, params)
```

## Câu hỏi 6: Tỉ lệ rời bỏ thay đổi thế nào theo từng giá trị?
Đổi tên cột trong `feature` để xem cột khác (ví dụ `subscription_type`, `number_of_profiles`, `payment_method`).
Cột số được chia thành 5 nhóm đều nhau.

```python
rates = churn_rate_by_feature(df, params)
rates[rates["feature"] == "last_login_days"]
```

```python
rates[rates["feature"] == "number_of_profiles"]
```
